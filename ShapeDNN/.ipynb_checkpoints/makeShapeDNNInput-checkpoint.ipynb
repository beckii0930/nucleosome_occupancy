```python
import numpy as np;
import time
import scipy.io
import sys
import math

def readInputAsArray(fileName):
    with open(fileName, 'r') as myfile:
        data = myfile.readlines()

    # Strip newline
    for i in range(0, len(data)):
        data[i] = data[i].rstrip()
    # print(data)
    return data
```

```python
def my_print(name, val):
    print(f'{name}: {val}')

# test=np.array([10])
# my_print('test', test)
```

```python
def list_str_to_float(arr):
    out = []
    for i in arr:
        out+=[float(i)]
    return out
```

```python
def append_list_to_np_arr(lst, np_arr):
    if np_arr.shape[0] == 0:
        np_arr = np.array([lst]) 
    else:
        np_arr = np.concatenate((np_arr, [lst]), axis=0) 
    return np_arr
```
