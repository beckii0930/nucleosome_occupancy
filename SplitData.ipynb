```python
import pandas as pd
import numpy as np
import scipy.io as sio
from os.path import dirname, join as pjoin
import os.path
import time
```

```python
# start = True
tic = time.perf_counter()
Edata_list = np.array([])
Ddata_list = np.array([])
for i in range(1, 25):
    print(i)
    fname = 'nucleosome_occupancy_' + str(i) + '.mat'
    if os.path.isfile(fname):
#         print("exist")
        mat_fname = pjoin(fname)
        mat_contents = sio.loadmat(mat_fname)
        
        edata=mat_contents['EnrichedData']
        ddata=mat_contents['DepletedData']

        if edata.size >0:
            if Edata_list.size == 0:
                print("Init Elist")
                Edata_list = edata;
            else:
                Edata_list = np.concatenate((Edata_list, edata), axis=0)
          
        if ddata.size >0:

            if Ddata_list.size == 0:
                print("Init Dlist")
                Ddata_list = ddata;
            else:
                Ddata_list = np.concatenate((Ddata_list, ddata), axis=0)
    else:
#         print("not exist, below are subfiles")
        for j in range(1, 20):
#             print(j)
            fname = 'nucleosome_occupancy_' + str(i) + '_' + str(j) + '.mat'
            if os.path.isfile(fname):
                mat_fname = pjoin(fname)
                mat_contents = sio.loadmat(mat_fname)
                edata=mat_contents['EnrichedData']
                ddata=mat_contents['DepletedData']
                if edata.size >0:
                    if Edata_list.size == 0:
                        print("Init Elist")
                        Edata_list = edata;
                    else:
                        Edata_list = np.concatenate((Edata_list, edata), axis=0)

                if ddata.size >0:
                    if Ddata_list.size == 0:
                        print("Init Dlist")
                        Ddata_list = ddata;
                    else:
                        Ddata_list = np.concatenate((Ddata_list, ddata), axis=0)
print("Enriched list size")
print(Edata_list.shape)    
print("Depleted list size")
print(Ddata_list.shape)    
toc=time.perf_counter()
print(f"Create master list took {toc - tic:0.4f} seconds")
```
