```python
def readInputAsArray(fileName):
    with open(fileName, 'r') as myfile:
        data = myfile.readlines()

    # Strip newline
    for i in range(0, len(data)):
        data[i] = data[i].rstrip()
    # print(data)
    return data
```

```python
def getmetrics(test_out_fn):
    test=readInputAsArray(test_out_fn)    
    true = []
    pred = []

    for i in test:
        line = i.split()
        for k in line:
            score = k.split("=")
            if score[0]=="TrueLabel":
    #             print(score[1][1])
                true += [int(score[1][1])]
            if score[0]=="Predicted":
    #             print(score[0][1])
                if (float(score[1][1:len(score[1])-1]) > 0.5):
                    pred += [1]
                else:
                    pred += [0]
    from sklearn.metrics import accuracy_score
    acc = accuracy_score(true, pred) 
    from sklearn.metrics import precision_score
    prec = precision_score(true, pred)
    from sklearn.metrics import recall_score
    rec = recall_score(true, pred)
    out = [acc, prec, rec]
    return out
```

```python
## Plot test output data
import numpy as np
import seaborn as sns

All_metrics = []
for i in range(1,11):
    test_out = '/home/yibei/Projects/nucleosome_occupancy/tbinet_stuff/test_dense10_kernel10_run'+ str(i) +'.out'
    out = getmetrics(test_out)
#     print(out)
    if len(All_metrics) == 0:
        if out[1] > 0.0 and out[2] > 0.0:
            All_metrics = [out]
    else:
        if out[1] > 0.0 and out[2] > 0.0:
            All_metrics = np.concatenate((All_metrics, [out]), axis=0)
print(len(All_metrics))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(All_metrics, columns = ['accuracy_score','precision_score','recall_score'])
print(df)
fig, ax = plt.subplots()
g = sns.lineplot(data=df, ax=ax)
ax.set_ylim(0.92, 1)
ax.set_xlabel('Model ID')
ax.set_ylabel('Metrics Score')
ax.set_title("Metrics among Trained Models")
```
