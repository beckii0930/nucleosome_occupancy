```python
import time
listone = [1, 2, 3]
listtwo = [4, 5, 6]
```

```python
tic = time.perf_counter()
joinedlist = listone + listtwo
toc=time.perf_counter()

print(f"Create master list took {toc - tic:0.4f} seconds")
joinedlist
```

```python
tic = time.perf_counter()
joinedlist = [listone] + [listtwo]
toc=time.perf_counter()

print(f"Create master list took {toc - tic:0.4f} seconds")
joinedlist
```
