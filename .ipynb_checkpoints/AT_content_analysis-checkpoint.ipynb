```python
import numpy as np
import pandas as pd
data_array = np.load('counts_50.npy')
arr_deplete = []
for i in data_array:
    arr_deplete.append(int(i))
arr_deplete =np.array(arr_deplete)
# print("\nData summary:\n", arr)
# print("\nData shape:\n", arr.shape)

data_array = np.load('counts_51.npy')
arr_enrich = []
for i in data_array:
    arr_enrich.append(int(i))
arr_enrich =np.array(arr_enrich)
# print("\nData summary:\n", arr_deplete)
print("\nData shape:\n", arr_deplete.shape)
# print("\nData summary:\n", arr_enrich)
print("\nData shape:\n", arr_enrich.shape)
```

```python
# fileName = "/home/yibei/Projects/nucleosome_occupancy/mc_pentamer_06042020.txt"
def readInputAsArray(fileName):
    with open(fileName, 'r') as myfile:
        data = myfile.readlines()

    # Strip newline
    for i in range(0, len(data)):
        data[i] = data[i].rstrip()
    # print(data)
    return data
def read_SD_table(fileName):
    tb_data = readInputAsArray(fileName)
    tb = {}
    DNAshape = tb_data[0].split("\t")
    # print(DNAshape)
    for i in range(1, len(tb_data)):
        line = tb_data[i].split("\t")
        key = line[0]
        value = line[1:]
        for j in range(0, len(value)):
            value[j] = float(value[j])
        tb[key] = value
    # print(tb)
    return(tb)
fileName = "/Users/yibeijia/Downloads/ProjectStuff/pentamer_table.txt"
fileName = "/Users/yibeijia/Downloads/ProjectStuff/mc_pentamer_06042020.txt"
tb = read_SD_table(fileName);
```

```python
df = pd.DataFrame(arr_deplete, columns = ['deplete'])
df['enrich'] = pd.DataFrame(arr_enrich)
df['kmer']=tb.keys()
df
```

```python
def min_max_norm(arr):
    new_arr = []
    min_v = min(arr)
    max_v = max(arr)
    for num in arr:
        new_arr=new_arr+[(num-min_v)/(max_v-min_v)]
    return(new_arr)
arr_enrich_norm =min_max_norm(np.array(arr_enrich))
arr_deplete_norm =min_max_norm(np.array(arr_deplete))
df_norm = pd.DataFrame(arr_deplete_norm , columns = ['deplete'])
df_norm ['enrich'] = pd.DataFrame(arr_enrich_norm )
df_norm ['kmer']=tb.keys()
df_norm 
```

```python
def AtrackCalc(seq):
    longest=''
#     curr_seq=''
    AllAtracks=[]
    start=0
    end=0
    count=0
    for letter in seq[1:]:
        count+=1
#         print(">>>>>>>>>>>>>")
#         print(f"current pos: {count}")
#         print(f"current letter: {letter}")        
#         print(f"last letter: {seq[count-1]}")        

        if seq[count-1] != letter: #If not equal, check sequence
            if seq[count-1]=='A' and letter=='T': #only ApT doesn't breaks Atrack\
#                 print("***ApT")
                end+=1
            else:
#                 print("A track broken")
#                 print(f"Appednign start:{start}, end:{end+1}")
                AllAtracks.append(seq[start:end+1])
                start=count
                end=count
        else: #If equal, expand sequence
#             print("equal, expand seq")
            end+=1
#         print(f"end: {end}")
        if end == len(seq)-1:
            AllAtracks.append(seq[start:end+1])
    for atrack in AllAtracks:
        if len(atrack) > len(longest):
            longest=atrack
#             longest = seq[start:end]
    return longest,AllAtracks

# longest,AllAtracks = AtrackCalc('TTAATTAATT') 
# # AllAtracks = AtrackCalc('AATTAC') 
# print(AllAtracks)
# print(longest)
def A_count(seq):
    out=0
    for letter in seq:
        if letter == 'A' or letter =='T':
            out+=1
    return out;
def GC_count(seq):
    out=0
    for letter in seq:
        if letter == 'G' or letter =='C':
            out+=1
    return out;
Acounts = []
GCcounts=[]
Atrack = []
for i in tb.keys():
    longest,AllAtracks = AtrackCalc(i)
    Atrack.append(len(longest))
    Acounts.append(A_count(i))
    GCcounts.append(GC_count(i))
df_norm['Atrack'] = Atrack
df_norm['Acounts'] = Acounts
df_norm
```

```python
df['Atrack'] = Atrack
df['Acounts'] = Acounts
df['GCcounts'] = GCcounts
```

```python
import seaborn as sns; 
from matplotlib import pyplot as plt

sns.set_theme(color_codes=True)
plt.figure(figsize=(15,8))
ax=sns.relplot(x="deplete", y="enrich", data=df, hue='Acounts', height=6, aspect=1.8)
# sns.lmplot(x="deplete", y="enrich", data=df_norm, height=5, aspect=2)
plt.plot([0, 1000000], [0,1000000], linewidth=2)
ax.fig.suptitle('Color by Total A counts')
# ax.set(xticks=[i for i in range(0, max(df['x-axis']) + 1, 4)],
#        yticks=[i for i in range(0, max(df['y-axis']) + 0, 1)])
```
